```python
def normalisasi(t):
    t = re.sub(r"[\u064B-\u0652\u0670]", "", t)  # hapus harakat
    t = re.sub(r"[أإآ]", "ا", t)                  # seragamkan alif
    t = re.sub(r"ة\b", "ه", t)                    # ta marbutah di akhir kata
    t = re.sub(r"ى\b", "ي", t)                    # alif maqshurah
    t = re.sub(r"\s+", " ", t)                    # rapikan spasi
    return t.strip()

print(normalisasi(hadis_18_lengkap))
```

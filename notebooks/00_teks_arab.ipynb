```python
import re

hadis_18 = "اتَّقِ اللهَ حَيْثُمَا كُنْتَ، وَأَتْبِعِ السَّيِّئَةَ الْحَسَنَةَ تَمْحُهَا، وَخَالِقِ النَّاسَ بِخُلُقٍ حَسَنٍ"

tanpa_harakat = re.sub(r"[\u064B-\u0652\u0670]", "", hadis_18)

print(hadis_18)
print(tanpa_harakat)
print("Jumlah karakter dengan harakat:", len(hadis_18))
print("Jumlah karakter tanpa harakat:", len(tanpa_harakat))
```
